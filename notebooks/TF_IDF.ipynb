- TF (term frequency): a word is important in a doc if it appears a lot in that doc.

- IDF (inverse document frequency): a word is less informative if it appears in many documents (“the”, “and”, etc.).

- TF-IDF vectors represent each doc/query as weighted word counts.

- Then you retrieve by cosine similarity between query vector and doc vector.

**In code:** `TfidfVectorizer` builds these vectors; `cosine_similarity` ranks documents.
