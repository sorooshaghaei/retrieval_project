BM25 (why it often beats TF-IDF)

BM25 is still keyword-based, but it improves TF-IDF by:

- saturating term frequency (100 occurrences isn’t 100× better than 1)

- normalizing by document length (long docs shouldn’t win just because they have more words)

**In code:** `BM25Plus` is a BM25 variant; `get_scores()` returns one score per document.
